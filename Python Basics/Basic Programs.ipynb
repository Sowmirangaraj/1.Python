## Python Code

## 1. Fibonacci Series

```python
n = 10
a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

## 2. Swap Two Numbers

```python
a = 10
b = 20

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

## 3. Prime Number Checker

```python
n = 17

if n < 2:
    print("Not a prime number")
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

## 4. Palindrome Checker

```python
word = "madam"

if word == word[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

## 5. Factorial of a Number

```python
n = 5
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print("Factorial:", factorial)
```

## 6. Pyramid Pattern

```python
rows = 5

for i in range(1, rows + 1):
    spaces = rows - i
    stars = 2 * i - 1
    print(" " * spaces + "*" * stars)
```

## 7. Reverse a String

```python
text = "Python"

reversed_text = text[::-1]

print("Original:", text)
print("Reversed:", reversed_text)
```

## 8. Find the Largest Number in a List

```python
numbers = [12, 45, 7, 89, 34]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest number:", largest)
```
